### 🚀 Step 5: Multi-Week API Data Ingestion & Dataset Generation

The next step scales up the pipeline to fetch comprehensive real-world observational data across multiple target windows:

* **Multi-Week Ingestion:** Systematically query NASA's API across multi-week time spans (~1.5 months) to gather a representative sample of close-approach events.
* **Automated Record Normalization:** Streamline raw JSON payloads into structured dictionaries, converting unit types (e.g., miss distance to Lunar Distances) on the fly.
* **Production Dataset Compilation:** Construct a clean, fully populated dataset ready for feature scaling, target flag assignment, and pipeline export.

```python
import random
from pathlib import Path

# المسار وقوائم التلويث
OUTPUT_PATH = Path("data/raw/ground_station_log.csv")
FIELDNAMES = ["neo_id", "observatory_code", "confidence_score"]
OBSERVATORY_CODES = ["G96", "703", "I41", "F51", "W88", " G96", "UNKNOWN"]

DROP_RATE = 0.10
GHOST_RATE = 0.10
DIRTY_RATE = 0.15

def _dirty_confidence_score():
    clean = f"{random.uniform(0.50, 0.99):.2f}"
    if random.random() < DIRTY_RATE:
        return random.choice(["", "N/A", "null", f"{clean} "])
    return clean

def _fabricate_ghost_id(used_ids):
    while True:
        candidate = str(random.randint(1_000_000, 4_999_999))
        if candidate not in used_ids:
            return candidate

def generate_sentinel_log_notebook(asteroids_list, output_path=OUTPUT_PATH):
    output_path.parent.mkdir(parents=True, exist_ok=True)
    ids = [str(ast.get("id")).strip() for ast in asteroids_list if ast.get("id")]

    n_drop = round(len(ids) * DROP_RATE)
    dropped = set(random.sample(ids, n_drop)) if n_drop else set()
    surviving_ids = [i for i in ids if i not in dropped]

    n_ghost = round(len(ids) * GHOST_RATE)
    used_ids = set(ids)
    ghost_ids = []
    for _ in range(n_ghost):
        ghost = _fabricate_ghost_id(used_ids)
        used_ids.add(ghost)
        ghost_ids.append(ghost)

    all_ids = surviving_ids + ghost_ids
    random.shuffle(all_ids)

    rows = [
        {
            "neo_id": neo_id,
            "observatory_code": random.choice(OBSERVATORY_CODES),
            "confidence_score": _dirty_confidence_score(),
        }
        for neo_id in all_ids
    ]

    with output_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=FIELDNAMES)
        writer.writeheader()
        writer.writerows(rows)

    print(
        f"✅ تم إنشاء الملف بنجاح!\n"
        f"إجمالي الكويكبات المدخلة: {len(ids)}\n"
        f"سجلات الملف الناتج: {len(rows)} أسطر "
        f"({len(dropped)} تم حذفهم, {len(ghost_ids)} IDs وهمية تم حقنها)\n"
        f"المسار: {output_path}"
    )


generate_sentinel_log_notebook(all_asteroids)
```

```python
def clean_and_blend_data(raw_asteroids, log_file_path):
   
    asteroids_dict = {str(ast["id"]).strip(): ast for ast in raw_asteroids if ast.get("id")}
    
   
    magnitudes = [
        float(ast["absolute_magnitude_h"]) 
        for ast in raw_asteroids 
        if ast.get("absolute_magnitude_h") is not None
    ]
    magnitudes.sort()
    median_magnitude = magnitudes[len(magnitudes) // 2] if magnitudes else 0.0

    clean_master_records = []
    seen_ids = set()

    
    with open(log_file_path, mode="r", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            neo_id = row["neo_id"].strip()
            
           
            if neo_id in seen_ids or neo_id not in asteroids_dict:
                continue
            
            nasa_info = asteroids_dict[neo_id]
            
          
            approaches = nasa_info.get("close_approach_data", [])
            if not approaches:
                continue
            
            first_app = approaches[0]
            approach_date = first_app.get("close_approach_date")
            
            try:
                speed = float(first_app.get("relative_velocity", {}).get("kilometers_per_hour", 0))
            except (ValueError, TypeError):
                speed = 0.0
                
            try:
                miss_dist = float(first_app.get("miss_distance", {}).get("kilometers", 0))
            except (ValueError, TypeError):
                miss_dist = 0.0

            diameter_info = nasa_info.get("estimated_diameter", {}).get("kilometers", {})
            min_d = float(diameter_info.get("estimated_diameter_min", 0.0))
            max_d = float(diameter_info.get("estimated_diameter_max", 0.0))
            avg_d = (min_d + max_d) / 2.0
            
          
            abs_mag = nasa_info.get("absolute_magnitude_h")
            if abs_mag is None:
                abs_mag = median_magnitude
            else:
                try:
                    abs_mag = float(abs_mag)
                except (ValueError, TypeError):
                    abs_mag = median_magnitude

          
            raw_score = row.get("confidence_score", "").strip()
            if raw_score in ["N/A", "null", "", None]:
                clean_score = None
            else:
                try:
                    clean_score = float(raw_score)
                except ValueError:
                    clean_score = None
            
            clean_observatory = row.get("observatory_code", "").strip()

            
            miss_distance_lunar = miss_dist / 384400.0 if miss_dist > 0 else 0.0
            
           
            priority_watch = 1 if (max_d >= 0.14 and miss_distance_lunar <= 10) else 0


            size_to_dist_ratio = (max_d / miss_distance_lunar) if miss_distance_lunar > 0 else 0.0


            if miss_distance_lunar <= 5:
                approach_category = "very_close"
            elif miss_distance_lunar <= 20:
                approach_category = "close"
            elif miss_distance_lunar <= 60:
                approach_category = "moderate"
            else:
                approach_category = "distant"

            merged_record = {
                "asteroid_id": neo_id,
                "name": nasa_info.get("name"),
                "is_hazardous": nasa_info.get("is_potentially_hazardous_asteroid", False),
                "absolute_magnitude_h": round(abs_mag, 2),
                "estimated_diameter_min_km": round(min_d, 4),
                "estimated_diameter_max_km": round(max_d, 4),
                "avg_diameter_km": round(avg_d, 4),
                "close_approach_date": approach_date,
                "close_approach_speed_kph": round(speed, 2),
                "miss_distance_km": round(miss_dist, 2),
                "miss_distance_lunar": round(miss_distance_lunar, 4),
                "size_to_distance_ratio": size_to_dist_ratio,
                "approach_category": approach_category,
                "priority_watch": priority_watch,
                "observatory_code": clean_observatory,
                "confidence_score": clean_score
            }
            
            clean_master_records.append(merged_record)
            seen_ids.add(neo_id)


    if clean_master_records:
        ratios = [rec["size_to_distance_ratio"] for rec in clean_master_records]
        min_r, max_r = min(ratios), max(ratios)
        
        for rec in clean_master_records:
            if max_r != min_r:
                scaled = (rec["size_to_distance_ratio"] - min_r) / (max_r - min_r)
            else:
                scaled = 0.0
            rec["scaled_size_to_distance_ratio"] = round(scaled, 6)
            rec["size_to_distance_ratio"] = round(rec["size_to_distance_ratio"], 6)

    return clean_master_records
```

```python
from pathlib import Path

LOG_PATH = Path("data/raw/ground_station_log.csv")


cleaned_data = clean_and_blend_data(all_asteroids, LOG_PATH)


os.makedirs("data/processed", exist_ok=True)
processed_file_path = "data/processed/clean_data.csv"


fieldnames_master = list(cleaned_data[0].keys())

with open(processed_file_path, mode="w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames_master)
    writer.writeheader()
    writer.writerows(cleaned_data)

print(f"✅ تم التنظيف والدمج وحفظ الملف النهائي بنجاح!")
print(f"إجمالي الكويكبات المنظفة والمدمجة: {len(cleaned_data)}")
print(f"المسار النهائي: {processed_file_path}")


crosstab = {(True, 1): 0, (True, 0): 0, (False, 1): 0, (False, 0): 0}
for rec in cleaned_data:
    key = (bool(rec["is_hazardous"]), rec["priority_watch"])
    crosstab[key] += 1

print("\n--- 📊 نتائج الـ Validation Check ---")
print(f"اتفقا (خطير ومهم): {crosstab[(True, 1)]}")
print(f"اتفقا (غير خطير وغير مهم): {crosstab[(False, 0)]}")
print(f"ناسا خطير و Sentinel عادي: {crosstab[(True, 0)]}")
print(f"Sentinel مهم و ناسا غير خطير: {crosstab[(False, 1)]}")

n_total = len(cleaned_data)
n_flagged = sum(1 for rec in cleaned_data if rec["priority_watch"] == 1)
pct_reduction = (1 - (n_flagged / n_total)) * 100
print(f"\n📉 نسبة تخفيض عبء المراجعة (Workload Reduction): {pct_reduction:.2f}%")
```
